```python
ai_hiring['Hiring_Status'] = ai_hiring['HiringDecision'].map({
    0: 'Not Hired',
    1: 'Hired'})
```

```python
ai_hiring['Hiring_Status'].value_counts()
```

```python
ai_hiring['Hiring_Status'].value_counts().plot(
kind='bar', color=['green', 'red'])
plt.title('Hiring Decision Distribution')
plt.xlabel('Hiring Status')
plt.ylabel('Number of Candidates')
plt.show()
```

```python
hiring_rate = ai_hiring['HiringDecision'].mean() * 100
print("Overall Hiring Rate:", round(hiring_rate, 2), "%")
```

```python
ai_hiring['Gender_Name'] = ai_hiring['Gender'].map({
0: 'Female',
1: 'Male'})
```

```python
gender_hiring = ai_hiring.groupby('Gender_Name')['HiringDecision'].mean() * 100
print(gender_hiring.round(2))
```

```python
gender_hiring.plot(
kind='bar',
color=['pink', 'blue'])
plt.title('Hiring Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Hiring Rate (%)')
plt.show()
```

```python
education_map = {
1: 'High School',
2: "Bachelor's",
3: "Master's",
4: 'PhD'}
ai_hiring['Education_Name'] = ai_hiring['EducationLevel'].map(education_map)
```

```python
education_hiring = (ai_hiring.groupby('Education_Name')['HiringDecision'].mean() * 100)
print(education_hiring.round(2))
```

```python
education_hiring.plot(
kind='bar',color='skyblue')
plt.title('Hiring Rate by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Hiring Rate (%)')
plt.show()
```

```python
ai_hiring['Experience_Level'] = pd.cut(
ai_hiring['ExperienceYears'],
bins=[-1, 2, 7, 15],
labels=['Entry Level', 'Mid Level', 'Senior Level'])
```

```python
experience_hiring = (
ai_hiring.groupby('Experience_Level', observed=True)['HiringDecision'].mean() * 100)
print(experience_hiring.round(2))
```

```python
experience_hiring.plot(
kind='bar', color='skyblue')
plt.title('Hiring Rate by Experience Level')
plt.xlabel('Experience Level')
plt.ylabel('Hiring Rate (%)')
plt.show()
```

```python
interview_score = (ai_hiring.groupby('Hiring_Status')['InterviewScore'].mean())
print(interview_score.round(2))
```

```python
interview_score.plot(
kind='bar',color=['green', 'red'])
plt.title('Average Interview Score by Hiring Status')
plt.xlabel('Hiring Status')
plt.ylabel('Average Interview Score')
plt.show()
```

```python
skill_score = (ai_hiring.groupby('Hiring_Status')['SkillScore'].mean())
print(skill_score.round(2))
```

```python
skill_score.plot(
kind='bar',color=['green', 'red'])
plt.title('Average Skill Score by Hiring Status')
plt.xlabel('Hiring Status')
plt.ylabel('Average Skill Score')
plt.show()
```

```python
ai_hiring['Overall_Score'] = (
ai_hiring['InterviewScore'] +
ai_hiring['SkillScore'] +
ai_hiring['PersonalityScore']) / 3
```

```python
overall_score = (
ai_hiring.groupby('Hiring_Status')['Overall_Score'].mean())
print(overall_score.round(2))
```

```python
overall_score.plot(
kind='bar',color=['green', 'red'])
plt.title('Average Overall Assessment Score by Hiring Status')
plt.xlabel('Hiring Status')
plt.ylabel('Average Overall Assessment Score')
plt.show()
```

```python
ai_hiring['Age_Group'] = pd.cut(
ai_hiring['Age'],
bins=[19, 25, 30, 35, 40, 45, 50],
labels=['20-25', '26-30', '31-35', '36-40', '41-45', '46-50'])
```

```python
age_hiring = (
ai_hiring.groupby('Age_Group', observed=True)['HiringDecision'].mean() * 100)
print(age_hiring.round(2))
```

```python
age_hiring.plot(
kind='bar',
color='skyblue')
plt.title('Hiring Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Hiring Rate (%)')
plt.show()
```

```python
strategy_map = {1: 'Online Application',
                2: 'Employee Referral',
                3: 'Campus Recruitment'}
ai_hiring['Recruitment_Method'] = ai_hiring['RecruitmentStrategy'].map(strategy_map)
```

```python
strategy_hiring = (
ai_hiring.groupby('Recruitment_Method')['HiringDecision']
.mean() * 100)
print(strategy_hiring.round(2))
```

```python
strategy_hiring.plot(
    kind='bar',color='skyblue')
plt.title('Hiring Rate by Recruitment Strategy')
plt.xlabel('Recruitment Strategy')
plt.ylabel('Hiring Rate (%)')
plt.show()
```
